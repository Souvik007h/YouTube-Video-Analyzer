```python
from src.youtube import get_transcript
from src.chunker import create_chunks
from src.retriever import create_retriever
from src.qa_chain import create_qa_chain


url = "https://www.youtube.com/watch?v=ZVWg18AXXuE"

print("1. Getting transcript...")
transcript = get_transcript(url)

print("2. Creating chunks...")
chunks = create_chunks(transcript, max_chars=2000)

print("3. Creating retriever...")
retriever = create_retriever(chunks, k=3)

print("4. Creating QA chain...")
answer_question = create_qa_chain(retriever)

```

```python
print("5. Asking question...")

question = "What is MLOPS?"

result = answer_question(question)

print("\n" + "=" * 70)
print("FINAL ANSWER")
print("=" * 70)

print(result["answer"])

print("\n" + "=" * 70)
print("CLICKABLE SOURCES")
print("=" * 70)

for i, source in enumerate(result["sources"]):

    print(
        f"\nSource {i + 1}: "
        f"{source['start_timestamp']} - "
        f"{source['end_timestamp']}"
    )

    print(f"Watch: {source['url']}")
```
